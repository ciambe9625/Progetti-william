```python
import pandas as pd
import numpy as np
```

```python
path="C:\\Users\\CambeddaAndrea\\Downloads\\riposo.csv"

header= []
data= []
line_data=[]
with open (file=path) as f:
    file_content= f.read()
    for line in file_content.splitlines():
        if not header:
            header = line.split(sep=',')
        else:
            #data.append(line.split(sep=','))
            #non basta inserire i dati, dobbiamo sistemare la colonna delle date (attualmente un range)
            line_data = line.split(sep=',')
    
            #prendo la colonna della data (attualmente un range) e splitto per ottenere i valori di inizio e fine
            data_values = line_data[0] = line_data[0].split(sep='-')

    
            correct_date = []
            if len(line_data) > len(header): #e il numero di dati nella riga è maggiore del numero di colonne dell'header
                year = line_data[1].split(sep='-')[0]
                correct_date.append(year)
            else:
                correct_date.append(None)
    
            #aggiungo il valore di data inizio al set di dati
            correct_date.append(data_values[0])
    
            start_index = len(line_data) - len(header) + 1 #calcola l'indice di partenza per i dati corretti
            #aggiungo gli altri dati a partire dall'indice calcolato
            correct_date.extend(line_data[start_index:])
            data.append(correct_date)
    
header = ['year', 'start_date'] + header[1:]
print(len(header), len(data[0]))
    
```

```python
df_riposo = pd.DataFrame(data= data , columns= header)
df_riposo
```

```python
df_riposo['year'] = df_riposo['year'].astype('Int64')
df_riposo['year'] = df_riposo['year'].fillna(value=df_riposo['year'].max()+1)


dates= df_riposo['start_date'].str.split()

mesi= ("Gen", "Feb", "Mar", "Apr", "Mag", "Giu", "Lug", "Ago", "Set", "Ott", "Nov", "Dic")

nuove_date= []

for year, date in zip(df_riposo['year'], dates):
    mese_num = mesi.index(date[0]) + 1
    nuove_date.append(f"{year}-{mese_num:02d}-{date[1]}")
 
df_riposo['start_date'] = nuove_date
df_riposo.drop(columns=['year'], inplace=True)
```
